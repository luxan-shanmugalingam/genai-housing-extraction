```python
import pandas as pd
```

```python
df = pd.read_csv('df_houses_new.csv')
```

```python
df.head()
```

```python
df.drop(['location', 'City', 'State'], axis = 1)
```

```python
df['Fitness_Facilities'].value_counts()
```

```python
recode_map = {
    "not": False,
    "provided": True
}

# Apply the map to the 'Internet' column.
# The original values are replaced with the corresponding values from the map.
df['Parking'] = df['Parking'].map(recode_map)
df['Laundry'] = df['Laundry'].map(recode_map)
df['Dishwasher'] = df['Dishwasher'].map(recode_map)
df['Stainless_Appliances'] = df['Stainless_Appliances'].map(recode_map)
df['AC'] = df['AC'].map(recode_map)
df['Pool'] = df['Pool'].map(recode_map)
```

```python
recode_map = {
    "not": False,
    "included": True
}

# Apply the map to the 'Internet' column.
# The original values are replaced with the corresponding values from the map.
df['Internet'] = df['Internet'].map(recode_map)
df['Cable'] = df['Cable'].map(recode_map)
```

```python
recode_map = {
    "No Outdoor Space": False,
    ("Combined","Balcony","Patio", "Deck")  : True
}

# Apply the map to the 'Internet' column.
# The original values are replaced with the corresponding values from the map.
df['Outdoor_Spaces'] = df['Outdoor_Spaces'].map(recode_map)
```

```python
df_new = df.drop(['location', 'City', 'State', 'Flooring'], axis = 1)
df_new.head()
```

```python
recode_map = {
    "not": False,
    "available": True
}

# Apply the map to the 'Internet' column.
# The original values are replaced with the corresponding values from the map.
df_new['Fitness_Facilities'] = df_new['Fitness_Facilities'].map(recode_map)
```

```python
df_new.head()
```

```python
df_new = df_new[df_new['rental_type']=='Monthly']
df_new.head()
```

```python
df_new['city_clean'] = df_new['city_clean'].str.title()
df_new['state_clean'] = df_new['state_clean'].str.upper()
df_new.head()
```

```python
df_new.to_csv('housing.csv', index=False)
```
