```python
# Save the content of the 'body' column to a text file
with open("body_column_texts.txt", "w", encoding='utf-8') as f:
    for entry in df['body'].dropna().unique():
        f.write(entry.strip() + "\n\n" + "-"*80 + "\n\n")
```

```python
# Save the content of the 'body' column to a text file
with open("title_column_texts.txt", "w", encoding='utf-8') as f:
    for entry in df['title'].dropna().unique():
        f.write(entry.strip() + "\n\n" + "-"*80 + "\n\n")
```

```python
# Save the content of the 'body' column to a text file
with open("amenities_column_texts.txt", "w", encoding='utf-8') as f:
    for entry in df['amenities'].dropna().unique():
        f.write(entry.strip() + "\n\n" + "-"*80 + "\n\n")
```
