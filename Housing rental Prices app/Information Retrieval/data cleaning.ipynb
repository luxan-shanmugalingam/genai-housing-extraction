```python
import pandas as pd
import glob

# Find all the chunk CSV files in your output directory
all_files = ['/kaggle/input/data-chunks/structured_data_chunk_0.csv','/kaggle/input/data-chunks/structured_data_chunk_1.csv', '/kaggle/input/data-chunks/structured_data_chunk_2.csv', '/kaggle/input/data-chunks/structured_data_chunk_3.csv', '/kaggle/input/data-chunks/structured_data_chunk_4.csv', '/kaggle/input/data-chunks/structured_data_chunk_5.csv', '/kaggle/input/data-chunks/structured_data_chunk_6.csv']


# Load each CSV into a DataFrame and store it in a list
li = []
for filename in all_files:
    df = pd.read_csv(filename, index_col=None, header=0)
    li.append(df)

# Concatenate all DataFrames into one
final_df = pd.concat(li, axis=0, ignore_index=True)

# Save the final, complete dataset
final_df.to_csv('/kaggle/working/final_comprehensive_dataset.csv', index=False)

print(f"Successfully combined {len(all_files)} chunks into one file with {len(final_df)} rows.")
display(final_df.head())
```
